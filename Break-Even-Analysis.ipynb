### Başabaş Noktası (Break-Even) Hesaplayıcı

Bir işletmenin ne kadar ürün satınca zarar etmeyi bıraktığını hesaplar.

- **Sabit maliyet (F):** Üretimden bağımsız giderler (kira, maaş vb.)
- **Birim fiyat (p):** Bir ürünün satış fiyatı
- **Birim değişken maliyet (c):** Bir ürünü üretmenin maliyeti

Başabaş miktarı: $Q^* = \dfrac{F}{p - c}$

Kâr: $\text{Kâr}(Q) = (p - c) \cdot Q - F$

```python
def break_even_quantity(fixed_cost: float, price: float, unit_cost: float) -> float:
    """
    Başabaş noktasındaki satış miktarını hesaplar.

    Args:
        fixed_cost (float): Toplam sabit maliyet.
        price (float): Birim satış fiyatı.
        unit_cost (float): Birim değişken maliyet.

    Returns:
        float: Kâr ve zararın sıfır olduğu satış miktarı.

    Raises:
        ValueError: Fiyat birim maliyetten büyük değilse.
    """
    margin = price - unit_cost
    if margin <= 0:
        raise ValueError("Fiyat birim maliyetten büyük olmalı, yoksa hiç kâr edilemez.")
    return fixed_cost / margin


def profit(quantity: float, fixed_cost: float, price: float, unit_cost: float) -> float:
    """Verilen satış miktarındaki kârı (negatifse zararı) hesaplar."""
    revenue = price * quantity
    total_cost = fixed_cost + unit_cost * quantity
    return revenue - total_cost

print("Fonksiyonlar başarıyla tanımlandı: break_even_quantity, profit")
```

### Kullanım Örneği

```python
# Örnek: Kira+maaş 10.000 TL, ürün 50 TL'ye satılıyor, üretimi 30 TL
fixed_cost = 10000
price = 50
unit_cost = 30

q_star = break_even_quantity(fixed_cost, price, unit_cost)
print(f"Başabaş miktarı: {q_star:.0f} adet (beklenen: 10000 / 20 = 500)\n")

for q in [300, 500, 800]:
    print(f"{q} adet satışta kâr: {profit(q, fixed_cost, price, unit_cost):,.2f} TL")
```

### Kullanıcıdan Girdi Alarak Hesaplama

```python
fixed_cost_input = float(input("Sabit maliyet (örn. 10000): "))
price_input = float(input("Birim satış fiyatı (örn. 50): "))
unit_cost_input = float(input("Birim değişken maliyet (örn. 30): "))
quantity_input = float(input("Planlanan satış miktarı (örn. 600): "))

q_star = break_even_quantity(fixed_cost_input, price_input, unit_cost_input)
p = profit(quantity_input, fixed_cost_input, price_input, unit_cost_input)

print("\n--- SONUÇ ---")
print(f"Başabaş miktarı: {q_star:.2f} adet")
print(f"{quantity_input:.0f} adet satışta {'kâr' if p >= 0 else 'zarar'}: {abs(p):,.2f} TL")
```
