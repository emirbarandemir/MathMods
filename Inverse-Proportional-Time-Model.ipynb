### Ters Orantılı Üretim Süresi Hesaplayıcı Fonksiyon

Bu fonksiyon, işçi (veya makine) sayısı ile bir işin tamamlanma süresi arasındaki ters orantılı ilişkiyi modeller. İşçi sayısı arttıkça sürenin azaldığı temel düşünceye dayanır.

```python
def calculate_production_time(
    initial_workers: float,
    initial_time: float,
    new_workers: float
) -> float:
    """
    İşçi sayısı ile üretim süresi arasındaki ters orantıyı hesaplar.

    Args:
        initial_workers (float): Başlangıçtaki işçi/makine sayısı.
        initial_time (float): Başlangıçtaki üretim süresi (örn. saat, gün).
        new_workers (float): Yeni işçi/makine sayısı.

    Returns:
        float: Yeni işçi/makine sayısına göre hesaplanan üretim süresi.

    Raises:
        ValueError: İşçi sayısı sıfır veya negatif girildiğinde.
    """
    if initial_workers <= 0 or new_workers <= 0:
        raise ValueError("İşçi sayısı sıfır veya negatif olamaz.")

    # Sabit iş yükünü hesapla (örn: işçi-saat)
    total_work_units = initial_workers * initial_time

    # Yeni işçi sayısına göre yeni süreyi hesapla
    new_time = total_work_units / new_workers
    return new_time

print("Fonksiyon başarıyla tanımlandı: calculate_production_time")
```

### Fonksiyonun Kullanım Örnekleri

Şimdi bu fonksiyonu test edelim.

```python
# Örnek 1: İşçi sayısı iki katına çıkarsa süre yarıya düşer.
initial_workers = 10
initial_time = 20.0 # 20 birim zaman (örn. saat)
new_workers = 20 # İşçi sayısı iki katına çıktı

new_time_1 = calculate_production_time(initial_workers, initial_time, new_workers)
print(f"Başlangıç: {initial_workers} işçi, {initial_time} saat. Yeni: {new_workers} işçi.")
print(f"Hesaplanan Yeni Süre: {new_time_1} saat (Başlangıç süresinin yarısı olması beklenir: {initial_time / 2.0})\n")

# Örnek 2: İşçi sayısı yarıya düşerse süre iki katına çıkar.
initial_workers_2 = 10
initial_time_2 = 20.0
new_workers_2 = 5 # İşçi sayısı yarıya düştü

new_time_2 = calculate_production_time(initial_workers_2, initial_time_2, new_workers_2)
print(f"Başlangıç: {initial_workers_2} işçi, {initial_time_2} saat. Yeni: {new_workers_2} işçi.")
print(f"Hesaplanan Yeni Süre: {new_time_2} saat (Başlangıç süresinin iki katı olması beklenir: {initial_time_2 * 2.0})\n")

# Örnek 3: Farklı bir işçi sayısı ile
initial_workers_3 = 8
initial_time_3 = 40.0
new_workers_3 = 12

new_time_3 = calculate_production_time(initial_workers_3, initial_time_3, new_workers_3)
print(f"Başlangıç: {initial_workers_3} işçi, {initial_time_3} saat. Yeni: {new_workers_3} işçi.")
print(f"Hesaplanan Yeni Süre: {new_time_3:.2f} saat")
```

### Kullanıcıdan Girdi Alarak Üretim Süresi Hesaplama

Şimdi kendi belirlediğimiz değerleri girerek üretim süresini hesaplayalım.

```python
# Kullanıcıdan girdi değerleri alalım
print("Lütfen üretim süresi hesaplaması için değerleri giriniz:")
initial_workers_input = float(input("Başlangıçtaki işçi/makine sayısı (örn. 10): "))
initial_time_input = float(input("Başlangıçtaki üretim süresi (örn. 20.0): "))
new_workers_input = float(input("Yeni işçi/makine sayısı (örn. 15): "))

# Kullanıcının girdiği değerlerle fonksiyonu çağır
user_calculated_time = calculate_production_time(
    initial_workers_input,
    initial_time_input,
    new_workers_input
)

print(f"\nKullanıcı Girdileri:")
print(f" Başlangıç İşçi/Makine Sayısı: {initial_workers_input}")
print(f" Başlangıç Süresi: {initial_time_input}")
print(f" Yeni İşçi/Makine Sayısı: {new_workers_input}")
print(f" Hesaplanan Yeni Süre: {user_calculated_time:.2f} birim zaman")
```
