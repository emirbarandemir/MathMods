```python
# Girdi al
a = float(input("Normal talep (a): "))
x = float(input("Yeni talep (x): "))
base_cost = float(input("Temel maliyet: "))
sensitivity = float(input("İlk türev (sensitivity): "))
curvature = float(input("İkinci türev (curvature): "))

# Hesapla
delta = x - a

linear_effect = sensitivity * delta
nonlinear_effect = (curvature / 2) * (delta ** 2)

predicted_cost = base_cost + linear_effect + nonlinear_effect

# Sonuç
print("\n--- SONUÇ ---")
print(f"Delta: {delta}")
print(f"Linear Effect: {linear_effect}")
print(f"Nonlinear Effect: {nonlinear_effect}")
print(f"Predicted Cost: {predicted_cost}")
```
